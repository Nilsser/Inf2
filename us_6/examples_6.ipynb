```python
import itertools

prefixes = prefixes = [
    'derm', 'derma', 'cuti', 'vuln', 'vult', 'trau', 'trauma', 'cica', 
    'les', 'dyn', 'dyna', 'dynam', 'vi', 'pot', 'tens', 'tenso', 
    'ton', 'tono', 'san', 'sana', 'cur', 'cura', 'act', 'actu', 
    'kin', 'cine', 'reg', 'regu', 'cyb', 'guber', 'con', 'syn', 
    'sym', 'co', 'peri', 'ambi', 'envi', 'enviro', 'meca', 'mech', 
    'circum', 'envic', 'hist', 'histo', 'margo', 'fiss', 'rhegm'
]

suffixes = [
    'act', 'ax', 'ex', 'it', 'is', 'os', 'on', 'um', 'mech', 'tron', 
    'nic', 'ics', 'stat', 'trop', 'ict', 'sia', 'sis', 'tion', 'ance', 
    'ol', 'in', 'an', 'as', 'ent', 'lytic', 'fix', 'ten', 'tract', 
    'hese', 'tense', 'ia', 'om', 'ora', 'ite', 'tice', 'seal', 
    'cure', 'lock', 'fuse', 'lytic'
]

# Generiert alle Kombinationen
names = [p + s for p, s in itertools.product(prefixes, suffixes)]
print(names)
```
