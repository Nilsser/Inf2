## Rekurisons Refresher

**iterative** Lösung Fakultät

```python
def it_factulty(n):
    factorial=1
    for i in range(1,n+1): #+1 wegen exklusiven ende
        factorial*=i
    return factorial
it_factulty(4)
```

```python
funktionen_im_stack = 0

def factorial(n):
    global funktionen_im_stack #nur fürs tracking, müsst nicht teil des kurses.
    funktionen_im_stack+=1
    print(f'es sind jetzt {funktionen_im_stack} calls offen')
    # 1. DER BASISFALL (Base Case)
    if n <= 1: #bekanntes erignis
        print(f"Basisfall erreicht: fakultät({n}) gibt 1 zurück.")
        return 1
    
    # 2. DER REKURSIONSSCHRITT (Recursive Step)
    else:
        print(f"Berechne: {n} * fakultaet({n-1})")
        ergebnis = n * factorial(n - 1)

        # Wenn der Basisfall erreicht wurde, "rollt" sich der Stapel von unten nach oben auf.
        funktionen_im_stack-=1
        print(f"Rückkehr aus dem Aufruf: fakultaet({n}) ist {ergebnis}", f'   berechnung erfolgte durch {n}*factorial({n-1})')
        print('es sind noch',funktionen_im_stack,'calls offen')
        return ergebnis

# Beispielaufruf
zahl = 3
print(f"Endergebnis von {zahl}! ist: {factorial(zahl)}")
```

## Dynamic Programming

```python
def fibonacci_iterative_table(n):
    if n <= 1:
        return n
    
    # Initialisierung der Tabelle (DP-Speicher)
    # Wir erstellen eine Liste der Länge n + 1
    dp = [0] * (n + 1)
    dp[1] = 1
    
    # Schrittweiser Aufbau der Lösung (Bottom-Up)
    for i in range(2, n + 1):
        dp[i] = dp[i-1] + dp[i-2]
        
    return dp[n]
```

wir brauchen aber eigentlich nur die letzten zwei Werte, (optimeirte Version)

```python
def fibonacci_optimized(n):
    if n <= 1:
        return n
    
    # Wir speichern nur die letzten zwei Werte statt einer ganzen Liste
    prev2 = 0  # fib(i-2)
    prev1 = 1  # fib(i-1)
    
    for _ in range(2, n + 1):
        current = prev1 + prev2
        # Werte für den nächsten Durchlauf verschieben
        prev2 = prev1
        prev1 = current
        
    return prev1
```
