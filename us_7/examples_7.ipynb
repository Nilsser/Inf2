```python
class Vehicle:
    def __init__(self, initial_position=0):
        """Initializes the vehicle at a given position (default is 0)."""
        self.position = initial_position

    def drive(self, duration):
        """Moves the vehicle based on a multiplier of 5."""
        self.position += 5 * duration
       


class RaceCar(Vehicle):
    def drive(self, duration):
        """Overrides the drive method to move at a multiplier of 10."""
        self.position += 10 * duration

car = Vehicle()
fast_car = RaceCar()
headstart_car = Vehicle(100)

car.drive(20)
fast_car.drive(12)
headstart_car.drive(-20)

print(car.position,fast_car.position,headstart_car.position)
print(headstart_car.position >= (car.position+fast_car.position)  )
```

```python
ls= [9,5,1,6,8]

def insertion_sort(items):
    n = len(items)
    for i in range(1, n):
        j = i
        # Wir tauschen das Element so lange nach links, 
        # wie das linke Element größer ist
        while j > 0 and items[j-1] > items[j]:
            # Das ist der "Bubble Swap"
            items[j], items[j-1] = items[j-1], items[j]
            j -= 1
        print(i,items)

print(insertion_sort(ls))
```

```python
def merge(a1, a2):
    """
    Führt zwei bereits sortierte Listen (a1 und a2) effizient 
    zu einer einzigen sortierten Liste zusammen.
    """
    # Initialisierung der Ergebnisliste b und der Zeiger i und j
    b, i, j = [], 0, 0
    # Solange in beiden Listen noch Elemente vorhanden sind, vergleichen
    while i < len(a1) and j < len(a2):
        if a1[i] < a2[j]:
            b.append(a1[i])
            i += 1
        else:
            b.append(a2[j])
            j += 1
            
    # Falls eine Liste fertig durchlaufen ist, werden die restlichen 
    # Elemente der anderen Liste (die bereits sortiert sind) angehängt.
    b += a1[i:]
    b += a2[j:]
    # print('s',b) #hier ausgeben lassen um zusehen wie die paare gebaut sind.
    return b


def merge_sort(a):
    """
    Sortiert eine Liste rekursiv nach dem Divide-and-Conquer-Prinzip.
    Zerlegt die Liste, bis sie nur noch aus Einzelteilen besteht, 
    und setzt sie sortiert wieder zusammen.
    """
    # Basisfall: Eine Liste mit 0 oder 1 Element ist bereits sortiert
    # print(a) Hier ausgeben lassen um zu sehen wie es tatsächlcih teilt.
    if len(a) <= 1:
        return a
    else:
        # Divide: Bestimmung der Mitte und Aufteilung in zwei Hälften
        # Rekursiver Aufruf für die linke und rechte Teilseite
        sorted_a1 = merge_sort(a[:len(a) // 2])
        sorted_a2 = merge_sort(a[len(a) // 2:])
        
        # Conquer: Die sortierten Teilstücke über die merge-Funktion vereinen
        return merge(sorted_a1, sorted_a2)

# Testlauf
test_liste = [2, 8, 5, 3, 9, 4, 1, 7]
print("Original:", test_liste)
print("Sortiert:", merge_sort(test_liste))
```
