## Matplotlib

```python
# datengenerierung
import random
import matplotlib.pyplot as plt
list_x = [random.randint(0, 20) for _ in range(20)]
list_y = [random.randint(0, 20) for _ in range(20)]
```

### scattering (Punkte)

```python
#list_x.append(50) # demo: x,y must have the same length, otherwise it will throw an error

#visualize the lists using matplotlib
plt.scatter(list_x, list_y,marker='x') # *** alternative markers: 'o', 'x', '^', 's', ***
plt.xlabel('List X')        # *** Achsenbeschriftung ***
plt.ylabel('List Y')        # *** Achsenbeschriftung ***
plt.title('Scatter Plot of Two Random Lists')  #*** Titel ***
plt.grid() 
plt.show()
```

### Plotting / Linie

```python
#line plot
plt.plot(list_x, list_y, marker='o')
plt.xlabel('List X')
plt.ylabel('List Y')
plt.title('Line Plot of Two Random Lists')
plt.grid()
plt.show()
```

## Färbung nach Kriterium

```python
#scatter plot with dots in different colors based on a criterion (e.g., above or below a certain threshold)
threshold = 1

# iterate over the paired values, not over enumerate; x and y are then numbers
colors = ['red' if y / x > threshold else 'blue' for x, y in zip(list_x, list_y)]

# komplizierter Fall: wir färben Punkte rot wenn y/x > als unser Schwellenwert ist,
# sonst blau; wir verwenden zip um über beide Listen gleichzeitig zu iterieren

#standardverfahren, plt scattering
plt.scatter(list_x, list_y, c=colors)#, c=colors die farben assignen
plt.xlabel('List X')
plt.ylabel('List Y')
plt.title('Scatter Plot with Color Based on Threshold')
plt.grid()
plt.show()
```

## Kombination mehrerer Plots

```python
#demo of a combination of line and scatter plot, using sine wave  and "noise"
import numpy as np
x = np.linspace(0, 10, 100) # 100 Werte von 0 bis 10
y_sin = np.sin(x) # Sinuswelle mit Rauschen
y_noise = np.random.normal(0, 0.2, size=x.shape) # Rauschen mit Mittelwert 0 und Standardabweichung 0.2 (normalverteiltes Rauschen)
y = y_sin + y_noise # Kombination aus Sinuswelle und Rauschen


# idealer sinus als linie, die punkte mit rauschen als scatter
#plot und scatter aufrufen um beide einzuzeichnen  
plt.plot(x, y_sin, label='Line Plot') # Linie zeichnen
plt.scatter(x, y, color='red', label='Scatter Plot') # Punkte zeichnen
#plt.plot(x, y_sin, label='another Line plot') #könnten auch einen weiteren plot hinzufügen


#plt suff
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Combination of Line and Scatter Plot')
plt.legend()
plt.grid()
plt.show()
```
