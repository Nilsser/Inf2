```python
# langsam, die funktion prüft auch dann noch,
# wenn sie schon festgestellt hat, dass die Liste nicht sortiert ist 
def is_sorted(liste):
    sorted_liste = sorted(liste)
    meets_condition = True
    for i in range(len(liste)):
        if liste[i] != sorted_liste[i]:
            meets_condition = False # hier sollte man bereits returnen! (oder break nach meets_condition = False)

    return meets_condition

# schneller
def is_sorted(liste):
    for i in range(len(liste) - 1):
        if liste[i] > liste[i + 1]:
            return False

    return True #müssen keine Variable einführen, da wir ja sofort returnen, wenn die Bedingung nicht erfüllt ist. 

# noch schneller
def is_sorted(liste):
    return sorted(liste) == liste

```

## Typischer, im autograder "unsichtbare" Fehler an der Prüfung!! 
anhand check_mountain in CX
Aufgabe: ein berg darf bis zum Gipfel nur ein steigendes Profil aufweisen, nach dem Gipfel nur ein fallendes.

```python
def check_mountain(mountain):
    max_height = max(mountain)
    is_ascending = True # Status-Flag: Befinden wir uns gerade in der Aufstiegsphase?

    prev_height = 0         # *** Entscheidende Stelle! ***
    
    for current_height in mountain:
        # 1. Zustandswechsel: Wenn wir bisher aufgestiegen sind, 
        # aber die aktuelle Höhe sinkt, beginnt die Abstiegsphase.
        if is_ascending and prev_height > current_height:
            is_ascending = False
        
        # 2. Validierung: Wenn wir uns im Abstieg befinden, 
        # darf kein Wert mehr größer sein als sein Vorgänger.
        if not is_ascending  and prev_height < current_height:
            return False, max_height #wie gelernt, direkt returnen
        
        # Aktuellen Wert für den nächsten Vergleich zwischenspeichern
        prev_height = current_height

    return True, max_height
```

**Wie könnte hier ein Problem entstehen?** 
* wir haben ground mit 0 definiert
* Grundsätzlich, wenn man von Bergen spricht, sinnvoll, (Meeresspiel=0m)
* Falls unsere Funktion sich aber in einem anderen Referenzsystem befindet, funktioniert die Funktion nicht mehr zwingend!!


&rarr; **Klassischer Fall, wie die initialisierung der Vergleichsvariable eine Rolle spielt!**


```python
# Vergleichswert mit erstem element intialisieren
def check_mountain_safe(mountain):

    if not mountain: # leere Liste, zur sicherheit prüfen
        return True, None
    

    max_height = max(mountain)
    is_ascending = True # Status-Flag: Befinden wir uns gerade in der Aufstiegsphase?

    prev_height = mountain[0]         # *** Entscheidende Stelle! ***
    
    for current_height in mountain:
        # 1. Zustandswechsel: Wenn wir bisher aufgestiegen sind, 
        # aber die aktuelle Höhe sinkt, beginnt die Abstiegsphase.
        if is_ascending and prev_height > current_height:
            is_ascending = False
        
        # 2. Validierung: Wenn wir uns im Abstieg befinden, 
        # darf kein Wert mehr größer sein als sein Vorgänger.
        if not is_ascending  and prev_height < current_height:
            return False, max_height #wie gelernt, direkt returnen
        
        # Aktuellen Wert für den nächsten Vergleich zwischenspeichern
        prev_height = current_height

    return True, max_height

beispiel_berg = [1,2,3,4,5,4,3,2,1]

#jedes element von beispiel_berg um 5 verkleinern
shifted_berg = [x - 5 for x in beispiel_berg]

#vergleich beider Funktionen mit beispiel_berg und shifted_berg
print(check_mountain(beispiel_berg)) 
print(check_mountain_safe(beispiel_berg))
print('\n Nun zum Problem: \n')
print(check_mountain(shifted_berg)) 
print(check_mountain_safe(shifted_berg)) 
```

weiteres, euch bereits bekanntes Beispiel: Eigene Maximum Funktion

```python
def maximum(values):
  """
  Diese Funktion ein Beispiel, wie man es KEINENFALLS implementieren sollte.
  Sie ist nicht robust, da sie davon ausgeht, dass alle Werte in der Liste kleiner als -99999 sind.
  """
  max = -99999
  for x in values:
    if(x > max):
      max = x;
  return max


def maximum(values):
  """
  Diese Funktion ist eine Verbesserung der vorherigen, da sie den Vergleichswert mit dem ersten Element der Liste initialisiert.
  Diese Funktion ist robust, da sie auch mit leeren Listen umgehen kann.
  """
  if(len(values) == 0):
    return 'None'
  
  max = values[0]
  for x in values:
    if(x > max):
      max = x;
  return max
```
