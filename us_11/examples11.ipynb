Gedankengang DP implementierung.

```python
#length:price
pricelist={
    0:0,
    1:1,
    2:5,
    3:8,
    4:9
}

def stab(n):

    #idee: eine 1-D tabelle zum ich von kurzen stäben zu langen vorzuarbeiten.
    table=[0]*(n+1) #n+1, da 0 indexiert

    #ich schribe alle preise werte in die dp tabelle

    for key,value in pricelist.items():
       
        if key<len(table):
            table[key]=value

    #jetzt bauen wir auf:
    # die längen 0 und eins sind eindeutig, die anderen können sich noch verändern. 
    #ich weiss bereits jetzt dass ich eine schleife benötige 
    #zum rest habe ich stand jetzt noch keinen ansatz.
   
    for length  in range(2,len(table)):
        #jetzt muss ich mir gedanken machen wie ich die tabelle von links nach rechts füllen kann.

        #ich nehme mir als anfangsbeispiel ein stab der länge zwei:
        # die frage für diesen stab lautet konkret, soll ich den stab halbieren um zwei mal ein stabe der länge 1 verkaufen
        #oder: einmal ein staab der länge 5 verkaufen. 
        
        # auch fäll mir ein: wenn ich einen einzigen schnitt setze, sind die entstehende stücke links in der tabelle,
        # daher kenne ich deren Lösung bereits. Dies stärkt die Vermutung, dass ich "mit bereits einem Schnitt setzen"
        # die bestsen optionen finden kann.
        
        #nächster schritt, Abstrahieren: muss ich mir immer diese Frage stellen (stab HALBIEREN oder so lassen?)
        # ich nutze ein anderer fall, bsp. länge 3: länge drei:
        #mir fällt auf, wenn ich ein stab der länge drei scchneide, passiert an beiden schnittstellen dasselbe (je ein stück der länge 1 und 2)
        # dies ist vielleicht ein einzelfall, nächstes beispiel: stab der länge 4
        # Ein Stab der Länge 4 kann ich mit einem schnitt entweder in die Längenkombninationen  (1,3) oder (2,2) teilen, dies ist spannend und scheint generll zu sein.      

        #jetzt versuche ich diese Frage stellung in den code zu bringen, ich weiss dass ich den max() nutzen kann um einfach aus möglichen optionen auszuwählen.

        #optionen "generieren"

        max_abschnitt_länge = length//2 +1 #meine Überlegung: wenn ich den schnitt über der Hälfte setze, dann mache ich wieder das selbe.
        
        optionen=[]
        for schnitt in range(0,max_abschnitt_länge+1): #hiermit prüfe ich alle möglichen positionen dieses schnitts.
            überbleibsel=length-schnitt
            preis=table[schnitt] #ich kann mir sicher sein dass wir diesen fall bereits anegschaut haben, und er drin ist.
            preis+= table[überbleibsel]
           
            optionen.append(preis)
        
        table[length]=max(optionen) #wir schreiben den wert in unsere liste:
        #ich merke dass ich eventuell damit bereits vorhandene werte überschriebe, dies ist mit unserem Ziel konfrom.
    return table[-1]
        
```
